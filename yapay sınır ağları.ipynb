```python
# =========================================================
# RF-DETR OBJECT DETECTION TRAINING
# FINAL GOOGLE COLAB VERSION
# =========================================================

# =========================================================
# 1. INSTALL LIBRARIES
# =========================================================

!pip install -q "rfdetr[train,loggers]"
!pip install -q roboflow supervision

# =========================================================
# 2. IMPORTS
# =========================================================

from roboflow import Roboflow
from rfdetr import RFDETRMedium

# =========================================================
# 3. DOWNLOAD DATASET FROM ROBOFLOW
# =========================================================

rf = Roboflow(api_key="7x4dQAq3o9ub8mwMNhhx")

project = rf.workspace("ekram-jvbia").project("coffeecup-s3qbh")

dataset = project.version(1).download("yolov11")

data_path = dataset.location

print("Dataset Path:", data_path)

# =========================================================
# 4. LOAD RF-DETR MODEL
# =========================================================

model = RFDETRMedium()

# =========================================================
# 5. TRAIN MODEL
# =========================================================

model.train(
    dataset_dir=data_path,
    epochs=50,              # Eğitim epoch sayısı
    batch_size=2,           # T4 GPU için güvenli değer
    grad_accum_steps=4,
    lr=1e-4,
    output_dir="rf_detr_output"
)

# =========================================================
# 6. FINISH
# =========================================================

print("RF-DETR eğitimi tamamlandı.")
print("Model ve sonuçlar rf_detr_output klasörüne kaydedildi.")
```

```python
# =========================================================
# RF-DETR RESULTS ANALYSIS
# TABLES + GRAPHS + METRICS
# =========================================================

# =========================================================
# IMPORTS
# =========================================================

import pandas as pd
import matplotlib.pyplot as plt

# =========================================================
# METRICS
# (Kendi eğitim sonuçlarından alındı)
# =========================================================

metrics = {
    "Metric": [
        "mAP50-95",
        "mAP50",
        "Recall",
        "Precision",
        "F1 Score"
    ],

    "Value": [
        0.3666,
        0.4389,
        0.6417,
        0.5694,
        0.5208
    ]
}

# =========================================================
# CREATE DATAFRAME
# =========================================================

df = pd.DataFrame(metrics)

print("\n========== RF-DETR METRICS TABLE ==========\n")

print(df)

# =========================================================
# BAR GRAPH
# =========================================================

plt.figure(figsize=(8,5))

plt.bar(df["Metric"], df["Value"])

plt.title("RF-DETR Performance Metrics")

plt.xlabel("Metrics")

plt.ylabel("Score")

plt.ylim(0,1)

plt.grid(True)

plt.show()

# =========================================================
# CLASS-BASED RESULTS
# =========================================================

class_metrics = {
    "Class": [
        "coffee-mug",
        "color-coffee-mug"
    ],

    "AP50-95": [
        0.4575,
        0.0737
    ],

    "Precision": [
        0.8889,
        0.2500
    ],

    "Recall": [
        0.5333,
        0.7500
    ]
}

class_df = pd.DataFrame(class_metrics)

print("\n========== CLASS METRICS ==========\n")

print(class_df)

# =========================================================
# CLASS GRAPH
# =========================================================

plt.figure(figsize=(8,5))

plt.bar(class_df["Class"], class_df["AP50-95"])

plt.title("Per-Class AP50-95")

plt.xlabel("Classes")

plt.ylabel("AP Score")

plt.ylim(0,1)

plt.grid(True)

plt.show()

# =========================================================
# FINAL MESSAGE
# =========================================================

print("\nRF-DETR sonuç analizi tamamlandı.")
```
